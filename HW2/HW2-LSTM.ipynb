```python
season = 12
test_size = 40

raw = dataset["S4248SM144NCEN"].values.astype("float64")
z = np.log(raw)
d = z[season:] - z[:-season]  # d[j] is for month j+12
print(len(raw), len(d))
```

```python
scaler = MinMaxScaler(feature_range=(0, 1))
scaler.fit(d[:-test_size].reshape(-1, 1))          # we should fit on train only
scaled_data = torch.from_numpy(scaler.transform(d.reshape(-1, 1))).to(dtype=torch.float32)
```

```python
train_set = scaled_data[:-test_size]
test_set = scaled_data[-test_size:]
```

```python
len(train_set), len(test_set)
```

```python
t0 = len(raw) - test_size            

plt.figure(figsize=(12,5))
plt.plot(range(t0), raw[:t0], color='blue', label='train')
plt.plot(range(t0, len(raw)), raw[t0:], color='red', label='test')
plt.legend(); plt.grid(True)
```

```python
def input_data(seq, ws):
    out = []
    L = len(seq)
    for i in range(L - ws):
        window = seq[i:i+ws]
        label = seq[i+ws:i+ws+1]
        out.append((window, label))
    return out
```

```python
window_size = 24

train_data = input_data(train_set, window_size)
```

```python
train_data[0]
```

### Construct  Model

```python
class LSTM(nn.Module):
    def __init__(self, input_size=1, hidden_size=64, dropout=0.2, output_size=1):
        super().__init__()
        self.hidden_size = hidden_size

        self.lstm = nn.LSTM(input_size, hidden_size, num_layers=1)
        self.linear = nn.Linear(hidden_size, output_size)
        self.hidden = (torch.zeros(1,1,self.hidden_size),
                       torch.zeros(1,1,self.hidden_size))

    def forward(self, seq):
        lstm_out, self.hidden = self.lstm(seq.view(len(seq),1,-1), self.hidden)
        pred = self.linear(lstm_out.view(len(seq),-1))
        return pred[-1]
```

### Train Model

#### Define Loss and Optimizer

```python
model = LSTM()
Loss = nn.MSELoss()
optimizer = torch.optim.AdamW(model.parameters(), lr=1e-3, weight_decay=1e-3)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, factor=0.5, patience=10)
```

```python
future = test_size

def predict_d(model, train_set, window_size, future):
    preds = train_set[-window_size:].view(-1).tolist()
    model.eval()
    with torch.no_grad():
        for f in range(future):
            seq = torch.FloatTensor(preds[-window_size:])
            model.hidden = (torch.zeros(1,1,model.hidden_size),
                            torch.zeros(1,1,model.hidden_size))
            preds.append(model(seq).item())
    model.train()
    return preds[window_size:]

def to_original(d_scaled):
    d_hat = scaler.inverse_transform(np.array(d_scaled).reshape(-1,1)).ravel()
    zz = list(z[:t0])
    for k in range(future):
        zz.append(zz[-season] + d_hat[k])           
    return np.exp(np.array(zz[t0:]))

real_test = raw[t0:]
rmse = lambda p: np.sqrt(np.mean((real_test - p)**2))
```

```python
epochs = 200
best_loss = float("inf")
L = []
model_path = f"{GLOB}/LSTM.pth"

for i in range(epochs):
    l = 0
    for seq, y_train in train_data:
        optimizer.zero_grad()
        model.hidden = (torch.zeros(1,1,model.hidden_size),
                        torch.zeros(1,1,model.hidden_size))
        y_pred = model(seq)
        loss = Loss(y_pred, y_train.view(-1))
        l += loss.item()
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), 1.0) 
        optimizer.step()
    print(f"epoch: {i} loss: {l:10.8f}")
    scheduler.step(l)
    L.append(l)

    # prediction on the test set in org scale
    preds = to_original(predict_d(model, train_set, window_size, future))
    test_loss = rmse(preds)
    print(f"          test RMSE: {test_loss:.1f}")

    if best_loss > test_loss:
        best_loss = test_loss
        checkpoint = {
            "model_state_dict": model.state_dict(),
        }
        torch.save(checkpoint, model_path)

    plt.figure(figsize=(12,4))
    plt.grid(True)
    plt.plot(range(t0-60, len(raw)), raw[t0-60:], label="original")
    plt.plot(range(t0, len(raw)), preds, label="prediction")
    plt.legend()
    plt.show()
print(best_loss)
```
